## 리스트(List)

변수에 값을 여러개 묶어서 넣고 싶음. 배열 형태로
순서가 있는 **"수정 가능한"** 객체의 집합
대괄호로 작성 되어지고 리스트 내부의 값은 콤마로 구분

```python
# 기본 구조
lst1 = []
print(lst1)
lst2 = [1,2,3] #문자(변수명)바로 뒤에 대괄호를 쓰면 인덱싱이고, 띄어쓰기가 되어 있으면 리스트이다.
print(lst2)
lst3 = ["My", "name"]
print(lst3)
lst4 = [1, 2, " My", "name"]#여기도 인덱스가 있다. 0123 네개가 있다. 리스트도 인덱싱 슬라이싱이 가능하다
print(lst4)
lst5 = [1, 2, ["My", "name"]]#리스트는 수정이 가능하기 때문에 현장에서 많이 쓴다(예로 퇴사자 등록 등등)
print(lst5)
```

```python
a = []  #공백가능
b= [1,2,3] #인트값 가능
c = ['My', 'name', 'is', 'jy'] #str문자값 가능
d = ['My', 'name', 1, 2] #int값 str값 둘다 가능
e =  ['My', 'name',[ 1, 2]] #리스트 안에 리스트 생성 가능 #컴퓨터는 문자를 읽을려고 하지않음. 약속된 언어가 아니기 때문 데이터로만 봄. 그래서 따옴표 붙혀서 꺼내주는거임. 숫자는 비교적 컴퓨터가 이해하기 쉬워서 그냥씀
```

### 리스트의 인덱싱, 슬라이싱

```python
lst1 = [2, 5, 7, 9, 10]
lst1

# []를 사용해서 인덱스 번호 1개에 접근하는 실습을 해보자
print("2에 접근 :", lst1[0])  #->대괄호가 문자 뒤에 바로 붙어있죠
print("7에 접근 :", lst1[2])
print("10에 접근 :", lst1[4]) #혹은 대괄호 안에 -1
print("lst1의 2번 요소와 4번 요소 덧셈 :", lst1[2] + lst1[4])
```

#### 리스트 인덱싱 예제

```python
lst2 = [1, 2, 3, ['a', 'b', 'c']]
lst2
# 문제 : 'b'라는 요소 하나에 접근하기(인덱싱 사용)
print("b에 접근 :", lst2[3][1])
```

```python
temp = lst2[3] #temp는 임시로 저장하는 변수
temp[1]
```

```python
print(temp)
print(lst2[3])
```

### 슬라이싱

```python
lst3 = [0, 1, 2, 3, 4, 5, 6]
lst3
```

```python
# 1 ~ 3까지 접근 해보자
print(lst3[1:4])
lst3[:5]
lst3[2:]
```

```python
#인덱싱과 슬라이싱의 차이
print(lst3[3])
print(lst3[3:4]) #리스트를 슬라이싱 하면 원래 가지고 있던 데이터인 리스트 형태가 유지가 된다.
```

```python
# 리스트 덧셈

lst1 = [1, 2, 3]
lst2 = [2, 3, 4]
lst1 + lst2 #여기서 알수있는건 리스트끼리 더하면 그냥 합쳐진다는거. 수식계산 x
```

# 튜플(tuple)

소괄호로 작성되어지고 수정이 불가능.

```python
tup1 = (1, 2, 3)
tup1
```

```python
tup1[2] #튜플도 인덱싱 슬라이싱이 가능하고 함수도 사용 된다
```

```python
len(tup1)
```

```python
tup2 = (3, 4, 5)  # 연산은 가능
tup1 + tup2
```

튜플은 함수안에 뭔가 추가 하거나 삭제 하는게 안된다. 그래서 실수를 하면 안되거나 절대 변경할일이 없는 데이터에 사용을 한다.
리스트는 요소가 몇개 들어갈지 명확하지 않은 경우에 사용하고 튜플은 요소의 개수를 사전에 정확히 알고 있을 경우에 사용한다
리스트는 가변적이며 튜플은 불변적이다
나중에 가다보면 리스트를 튜플보다는 많이 사용하게 된다.

in : 찾고자 하는 값이 포함되어 있으면 true 없으면 false
not in :찾고사 하는 값이 포함되어 있지 않으면 true

```python
lst4
print(6 in lst4)
print(6 not in lst4) #in은 반복문에서 굉장히 중요하게 사용이 된다
```

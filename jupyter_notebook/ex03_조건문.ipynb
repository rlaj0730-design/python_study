## 조건문(if elif else)

# if문

```python
if True :
    print("if문 안쪽 실행") #콜론 엔터만 쳐도 들여쓰기가 됨. 안쪽에 있으면 들여쓰기 #만약에 조건식이 참이면 안쪽, 거짓이면 바깥쪽을 실행해주세요
print("if문 바깥쪽 실행") #바깥쪽에 있으면 바깥쓰기
```

```python
if False :
    print("if문 안쪽 실행")  #만약에 조건식이 참이면 바깥쪽, 거짓이면 안쪽을 실행해주세요
print("if문 바깥쪽 실행")
```

## 조건문 elif(처음 조건식1이 false면 다음 조건식2을 확인하고 true면 실행문장2 실행)

중간에 elif(경우의수)를 여러개 써도 됨

```python
#elif문 예제
num1 = int(input("첫 번째 정수 입력 >>"))
num2 = int(input("두 번째 정수 입력 >>"))
# 1. num1 > num2
# 2. num1 < num2
# 3. num1 == num2
if num1 > num2:
    print("첫 번째 정수가 더 큽니다.")
elif num1 < num2:
    print("두 번째 정수가 더 큽니다.")
else:
    print("두 수가 똑같습니다.")
```

```python
score = int(input("점수 입력 >>"))
if score <= 100 and score >= 90:
    print(f"{score}점은 A학점 입니다")
elif score < 90 and score >= 80:
    print(f"{score}점은 B학점 입니다")
elif score < 80 and score >= 70:
    print(f"{score}점은 C학점 입니다")
elif score < 70 and score >= 60:
    print(f"{score}점은 D학점 입니다")
else:
    print(f"{score}점은 F학점 입니다") # 나중에 깨달은 건데 첫번째 if문 다음으로는 and 안써도 되었을거 같음. 최솟값만 설정해주면 위에서 순서대로 걸러주니까
```

```python
score = int(input("점수 입력 >>"))
if 90 <= score <= 100:
    print(f"{score}점은 A학점 입니다.")
elif 80 <= score:
    print(f"{score}점은 B학점 입니다.")
elif 70 <= score:
    print(f"{score}점은 c학점 입니다.")
elif 60 <= score:
    print(f"{score}점은 D학점 입니다.")
else:
    print(f"{score}점은 F학점 입니다.")
    
```

# 네이버 검색 자동화하기
1. 크롬 아이콘 클릭
2. 네이버 페이지 찾아가기
3. 검색창에 커서 올려놓기
4. 검색어 입력
5. 검색 버튼 클릭 or ENTER
6. 크롬 종료

```python
from selenium.webdriver.chrome.service import Service
# 사용자가 사용하는 최신버전의 크롬 드라이버와 일치하는 드라이버를 찾아주는 라이브러리
from webdriver_manager.chrome import ChromeDriverManager
# 최신버전의 크롬 드라이버를 찾고 다운로드 해주는 라이브러리
# ChromeDriverManager().install()  : 크롬드라이버 업데이트 및 설치
s = Service(ChromeDriverManager().install())
driver = wb.Chrome(service = s)
```

```python
from selenium import webdriver
# 크롬창 켜기
driver = wb.Chrome()

# 네이버 페이지 주소를 알려주지 않으면 빈 창이 뜨기 때문에
# 페이지 주소를 보내주자
driver.get('https://www.naver.com/')
```

```python
# 검색창의 위치 찾기
# driver.find_element  : 단수  --> 여러개 있더라도 맨 위에 있는 요소 하나만
# driver.find_elements : 복수 --> 리스트 안에 담겨서
# Selenium은 BeautifulSoup과 달리 실시간으로 변화하는 동적 웹페이지의 요소를 직접 제어할 수 있음
search = driver.find_element(By.ID, "query") # 실시간 화면에서 ID가 'query'인 검색창 요소 가져오기
search
```

```python
# 검색어 입력하기
search.send_keys("노윤서")
```

```python
# ENTER 키 입력 or 검색 버튼 클릭
# ENTER --> 모두 대문자로
search.send_keys(Keys.ENTER)
```

```python
# 현재 페이지에서 '이미지' 탭 클릭
# 전체 웹페이지의 정보를 모두 다루고 있는
tab = driver.find_elements(By.CSS_SELECTOR, "#lnb > div.lnb_group > div > div.lnb_nav_area._nav_area_root > div:nth-child(2) > a")
tab[0].click()
```

```python
# 크롬창 종료
driver.close()
```

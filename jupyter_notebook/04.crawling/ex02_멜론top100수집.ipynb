```python
# 브라우저인척 속이는 작업!
user_agent = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36'}
```

```python
#headers = 서버에게 보여주는 "내 웹 브라우저 신분증(또는 통행증)"이라고 표현할 수 있습니다.
#          레스토랑에 갈 때 '나는 옷을 제대로 차려입은 일반 손님이에요'라는 걸 보여주는 드레스 코드나 멤버십 카드 같은 역할이라고 생각하시면 편합니다.
res = req.get('https://www.melon.com/chart/index.htm', headers = user_agent)
res
```

```python
# 응답받은 데이터 파싱(=bs 객체화)
# 제목, 가수 수집해보기!
# 수집한 데이터를 tittle_list / singer_list 각각 담기!
# 각각 길이 확인하기!
melon = bs(res.text, 'lxml')
data = melon.select('div.ellipsis.rank01>span>a')
title_list = [data[i].text for i in range(len(data))]
data2 = melon.select('div.ellipsis.rank02>a')
singer_list = [data2[i].text for i in range(len(data2))]
singer_list
```

```python
#bs객체화 (민수쌤 풀이)
soup = bs(res.text, 'lxml') #파싱한 순간을 캡쳐해서 정보를 수집해옴
soup
```

```python
# 노래 제목 수집
song = soup.select('div.ellipsis.rank01>span>a')
# 선택자 이름 안에 공백은 자손선택자를 의미 -> 공백은 온점(.)으로 대체 해줘야 한다.
```

```python
# 1. .text와 .strip()의 차이
# - .text : HTML 태그를 제거하고 내부 텍스트만 추출 (단, 앞뒤/내부의 줄바꿈 '\n'이나 공백은 그대로 유지됨)
# - .strip() : 문자열의 양 끝(앞/뒤)에 존재하는 불필요한 공백, 줄바꿈, 탭 문자 등을 제거

# 2. 크롤링 시 .strip()을 항상 쓰는 이유
# - 눈에 보이지 않는 여백(\n, \t 등)으로 인한 데이터 오염 방지
# - 문자열 비교(==) 및 조건문 처리 시 공백 때문에 발생하는 오작동 방지
# - DB/엑셀 저장 전 깔끔한 데이터 전처리 효과

# 결론: 크롤링 시 텍스트 추출은 습관적으로 `element.text.strip()` 형태 사용 권장
```

```python
# 반복문 써서 전체 데이터 수집
song_lst = []
singer_lst = []
rank = []
for i in range(len(song)):
    song_lst.append(song[i].text.strip())
    singer_lst.append(singer_list[i].select_one('a').text)
    rank.append(i+1)
```

```python
# 딕셔너리 만들기
dic_melon = {"노래 제목" : song_lst, "가수" : singer_lst, "순위" : rank}

# 딕셔너리 -> DataFrame으로
import pandas as pd
df = pd.DataFrame(dic_melon)
df
```

```python
df.set_index("순위")  #순위를 제일 앞으로 보내겠습니다.
```

```python
# 1. BeautifulSoup의 유래
# - HTML 태그가 하나의 거대한 문자열로 뭉쳐있어 원하는 데이터를 찾기 어려움 (Tag Soup)
# - 이를 해결하기 위해 HTML 문자열을 의미 있는 객체(Tree 구조)로 파싱하여 정돈해 주는 도구가 BeautifulSoup

# 2. 파싱 시점(Capture)과 동적 데이터의 이해 (★동적 크롤링 핵심)
# - soup = bs(res.text, 'lxml')를 실행하는 '그 순간'의 HTML 상태를 캡쳐(정적 스냅샷)하여 가져옴
# - 따라서 페이지가 로딩된 후 JavaScript 등으로 실시간 변경되는 데이터(동적 페이지)는 
#   이 시점에 포함되지 않을 수 있으므로 주의 필요

# 3. HTML 계층 구조 (상위 태그 탐색)
# - 원하는 데이터(텍스트)가 위치한 태그에 셀렉터나 ID/Class가 명확하지 않다면,
#   F12(개발자 도구)를 통해 상위(부모) 태그로 올라가며 유일한 기준점을 찾아 접근함

# 4. CSS 선택자 작성 시 공백(' ') 대신 점('.')을 붙이는 이유
# - 공백(' '): 자손 선택자(Child/Descendant)를 의미함 -> 전혀 다른 하위 태그를 찾게 되어 오류 발생
# - 점('.'): 멀티 클래스(Multi-class) 지정 -> 한 태그가 여러 클래스를 동시 소지할 때 연결
#   ex) <div class="ellipsis rank01"> 
#       -> 'ellipsis'와 'rank01' 두 클래스를 모두 가진 태그를 지정하기 위해 'div.ellipsis.rank01'로 작성
```

```python
from selenium import webdriver as wb
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup as bs
```

```python
driver = wb.Chrome()
url = 'https://www.hsd.co.kr/menu/menu_list'
driver.get(url)
```

#### 동적 페이지 데이터 수집 방법
1. selenium 라이브러리를 사용하는 방법
2. BeautifulSoup 객체화 진행 후 수집하는 방법

```python
# 첫 번째 방법 사용
# 어떤 객체 값인지 눈으로 확인할수 없다. 동적 페이지를 가져올수 있다.
driver.find_elements(By.CSS_SELECTOR, "h4.h.fz_03")
```

```python
driver.find_elements(By.CSS_SELECTOR, "h4.h.fz_03")[0].text  #이렇게 하면 확인할 수는 있다.
```

```python
# 두 번째 방법 사용
# 어떤 객체 인지눈으로 확인 가능하다.
# driver.page_source -> html 문서를 문자열 형태로 return
soup = bs(driver.page_source, 'lxml')
menu = soup.select('h4.h.fz_03')
menu[0].text
```

```python
# 가격 수집
price = soup.select("div.item-price > strong")
price
```

```python
# 반복문 사용해서 이름과 가격 수집
menu_lst = []
price_lst = []
for i in range(len(menu)):
    menu_lst.append(menu[i].text)
    price_lst.append(price[i].text)
```

```python
len(menu_lst), len(price_lst)
```

#### 더보기 버튼 누르기
- 한솥도시락 페이지에서 더보기 버튼은 한 개만 존재한다.
- 더보기 버튼을 몇 번 눌러야 전체 메뉴가 보이는지 모른다는 가정 하에 코딩해보자.

```python
# 더보기 버튼 위치 찾기
btn_more = driver.find_element(By.CSS_SELECTOR, "a.c_05")
btn_more.click()
```

```python
# 더보기 버튼이 몇 번 중복되는지 알수 없으므로 20번 정도 눌러주자.
#근데 중간에 버튼이 없어지면 오류가 나면서 오류난 부분에서 멈춰버린다.
for i in range(20):
    btn_more = driver.find_element(By.CSS_SELECTOR, "a.c_05")
    btn_more.click()

print("더보기 버튼 클릭 완료!")
```

# 예외처리 하는방법

```python
# 원래 코드 실행 시 에러가 발생하면 밑에 작성된 코드들은 실행되지 않고 중간에 멈춤
# 멈추지 말고 계속 해서 아래 코드를 실행할 수 있도록 예외처리 진행을 해보자
# 트라이 익셉트 문을 쓰기 전에 밖에서 먼저 해보고 오류를 찾아보자.
try:
    # 에러가 나기 전까지 실행해야 할 문장이 들어가는 부분
    for i in range(20):
        btn_more = driver.find_element(By.CSS_SELECTOR, "a.c_05")
        btn_more.click()
except:
    # 만약 try부분을 실행하다가 오류가 난다면 except로 넘어와주세요
    print("더 보기 버튼 클릭 완료!")
```

#### time라이브러리
- 코드 중간에 쉬는 시간을 부여해주자
- 너무 빨리 버튼을 누르니까 페이지에 부하가 걸리기도 하고 사람이 아니라고 인식해서 접속을 막을 수도 있기 때문

```python
import time
# 사람이 처리하는 것처럼 해야 서버에 부담이 가지 않는다!
```

```python
try:
    # 에러가 나기 전까지 실행해야 할 문장이 들어가는 부분
    for i in range(20):
        btn_more = driver.find_element(By.CSS_SELECTOR, "a.c_05")
        btn_more.click()
        time.sleep(2) # 2초의 쉬는시간 부여
except:
    # 만약 try부분을 실행하다가 오류가 난다면 except로 넘어와주세요
    print("더 보기 버튼 클릭 완료!")
```

 -  크롤링 원칙 : 사람이 하는 행위를 그대로 코드로 구현한다. (웹브라우저를 들어가서 주소창에 주소 입력해서 서버에게 네이버 사이트를 주라고 요청. f12로 확인가능)
    1. 브라우저를 실행한다.

    3. 내용을 확인한다.

```python
# 도구 불러오기
import requests as req

# requests = 우리가 웹 브라우저(크롬 등) 주소창에 URL을 입력하고 엔터를 눌러 웹 페이지를 불러오는 행위를 코드(프로그램)로 똑같이 수행해 주는 도구입니다.(사이트의 요청과 응답 정보를 처리)
```

```python
# 1. 네이버 뉴스를 url로 "요청"하기
res = req.get('https://news.naver.com/')
res
# <Response [200]> : 통신에 성공 = 요청을 했고, 응답을 받았다. 200은 성공 응답상태 코드. 404처럼 서버가 요청받은 리소르를 찾을수없다는걸 의미하는 코드도 있다.
# 503 자주본다. 서버에 문제가 생겨서 서버가 요청을 처리할수 없을때.

# <Response [400]> : 400번대는 요청에서의 오류 = 사용자(client)가 잘못했거나(ex> urk 주소오타)
#                                              만료된 페이지

# <Response [500]> : 500번대는 응답에서의 오류 = 서버측의 오류

# https://developer.mozilla.org/ko/docs/Web/HTTP/Reference/Status --- 코드 참고.
```

```python
# 2. 응답받은 데이터에서 원하는 정보를 추출
res.text

# res.text : 응답 데이터를 문자열 한줄로 보여주는 속성 - 맨 앞을 보면 됨. 따옴표가 있으니 문자열임.
# -> html 문서를 반환
```

```python
#!pip install BeautifulSoup4    #재료 많은 수프에서 재료 골라 떠먹는데에서 유래
```

```python
# bs4 => 문자열을 그대로 html 문서로 만들어서 적용. 위에 리퀘스트로 반환받은 정보는 보기에 너무 복잡하니까

from bs4 import BeautifulSoup as bs
# 전문용어로 '문서를 파싱'해서 특정 태그를 쉽게 찾을 수 있도록 도와주는 라이브러리 이다.
# 파싱 : 비정형화 되어있는 순수 문자열 데이터를 html 코드로 인식될 수 있게끔 하는 작업을 의미한다.
```

```python
# !pip install lxml
```

```python
# bs4라이브러리 사용방법 : bs(문자열 데이터, 파싱방법)  --> 위에서 as bs했으니까
soup = bs(res.text, 'lxml') # html.parser, html5lib 라는 방법도 있다. 근데 lxml이 좀 빠르다.
soup
```

### 선택자 복습
1. 요소 선택자 : 태그 그 자체 ex> span, a, p, strong, div, ... 등등
2. 클래스 선택자 : 클래스를 통해서 선택 ex> .classname
3. 아이디 선택자 : 아이디를 통해서 선택 ex> #idname
4. 자식 선택자 : 특정 요소의 자식 선택 ex> div>a

```python
# select() : 요소 지정, css 선택자 활용
data = soup.select('span.Nitem_link_menu') # 개발자 창 들어가서 찍어보고 더블클릭으로 가져옴
data
# 선택자를 활용해서 자세하게 적어줘야 원하는 요소(태그)를 지정할 수 있다!
```

```python
# 위에 보면 리스트 형식이니까 인덱싱도 가능. 
data[1]
```

```python
# content(내용) 하나 선택해서 추출!
data[5].text
```

```python
# content 여러개 추출!
# content를 지정하기 위해서 태그명을 제거하는 과정이 .text 였다.
# 인덱싱 + text속성으로 작업 + 반복문
for i in range(len(data)):
    print(data[i].text)
```

```python
# 뭐가 들어있는지 모를떄 반복문에 찍어보면 됨.
for d in data:
    print(d.text)
```

```python
# 리스트를 선언하고 content만 담아두기
news_item = []
for i in range(len(data)):
    # 리스트에 값을 하나씩 추가하는 함수 #내가 답변한 대입연산자는 변수 값을 0으로 지정해놓고 숫자를 계속 더할때 사용했었음..
    news_item.append(data[i].text)

news_item
```

```python
# 리스트 컴프리헨션 : 리스트에 값을 쉽게 추가하는 방법. 바로 위 셀처럼 3줄 안써도 됨
# 변수명 = ['넣고자 하는 데이터' 'for문의 첫줄']
news_item1 = [data[i].text for i in range(len(data))]
news_item1
```

```python
# 실습 : 네이버에서 원하는 키워드를 검색하고 뉴스 탭에서 뉴스 타이틀 수집!
# 1. url 요청
# 2. 응답 받은 정보를 파싱
# 3. 뉴스 타이틀 찾아서 지정(태그, 선택자 활용), 가져오기
# 4. content만 가져와서 리스트로 담기!

# 3번에서 클래스 선택자로 클래스 속성 가져올때 공백 -> . 으로 교체

lot = req.get('https://search.naver.com/search.naver?ssc=tab.news.all&where=news&sm=tab_jum&query=%EB%A1%AF%EB%8D%B0%EC%9E%90%EC%9D%B4%EC%96%B8%EC%B8%A0')
giant = bs(lot.text, 'lxml')
data2 = giant.select('span.sds-comps-text.sds-comps-text-ellipsis.sds-comps-text-ellipsis-1.sds-comps-text-type-headline1')
lotte = [data2[i].text for i in range(len(data2))]
lotte

# 1번 유알엘 요청해서 쿼리 뒤에 내용 변경 하면 그 내용으로 데이터 불러와짐.
# 2번 에서 텍스트쓰면 태그도 같이 나오고 4번에서는 텍스트 쓰면 콘텐츠만 출력되는데
# 그이유는 2번에서는 원시데이터를 불러온 상태라 태그를 태그로 인식 못하고 하나의 문자열로 인식해서 다출력함.
```

```python
# 주소내용 쿼리 뒤에 포맷팅 사용해서 검색어를 입력하면 자동으로 뉴스 기사 제목들 출력되게 만들수도 있음
user_input = input("검색어를 입력하세요 >>")
lot = req.get(f'https://search.naver.com/search.naver?ssc=tab.news.all&where=news&sm=tab_jum&query={user_input}')
giant = bs(lot.text, 'lxml')
data2 = giant.select('span.sds-comps-text.sds-comps-text-ellipsis.sds-comps-text-ellipsis-1.sds-comps-text-type-headline1')
lotte = [data2[i].text for i in range(len(data2))]
lotte
```

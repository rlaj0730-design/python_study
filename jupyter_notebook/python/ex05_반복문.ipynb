## while문
- 조건식이 True인 동안 실행문장을 반복해서 실행
- 몇 번 반복할지 정해져 있지 않은 경우 사용한다.

```python
num = 1
while num <= 3:
    print(num)
    num += 1
```

```python
#while문을 사용해서 " 파이썬 최고!!"를 10번 출력하시오.
num = 1
while num <= 10:
    print("파이썬 최고!!")
    num += 1
    
#절대 절대 무한 루프에 빠지면 안됨. 만약 빠지면 위에 네모 표시 누르면 됨
```

#### while문에서 무한반복을 피하는 방법

```python
num = 1
while True:
    print(num)
    num += 1
    if num > 3:    # if break
        break

#break는 어떤 상황이든 코드의 흐름이 break를 만나는 순간 반복문은 무조건 종료된다.
```

## for문

이해하기 엄청 어려움. 반복끝나면 알아서 나가기 때문에 브레이크 걸 필요 없다

for 변수 in 문자열(or 리스트 or 튜플):
    print(변수)

- 반복 횟수가 정해져있을 때 사용
- 보통 인덱스가 있는 리스트, 튜플, 문자열과 함께 사용된다

```python
#for문 예시
hi = "안녕하세요" # 안, 녕, 하, 세, 요 프린트 하나씩 하는거랑 같음
for h in hi:     #숫자로 부르는거보다 이름으로 바로 부르는게 낫기 때문
    print(h)
    
```

```python
lst_food = ['햄버거', '치킨', '피자']
for food in lst_food:
    print(f"{food}씨~ 상담합시다 상담실로 오세요~~") #문자 보낼때 이름만 바꿔서 보낼때 등 사용
```

```python
num = 1
score_list = [90, 45, 70, 60, 55]
for score in score_list:
    if score >= 60:
        print(f"{num}번 학생은 합격입니다")
    else:
        print(f"{num}번 학생은 불합격입니다")
    num += 1
    
    #내가 푼 풀이
```

```python
score_list = [90, 45, 70, 60, 55]
cnt = 0
for score in score_list:
    cnt +=1
    if score >= 60:
        print(f"{cnt}번 학생은 합격입니다.")
    else:
        print(f"{cnt}번 학생은 불합격입니다.")  #선생님 풀이
```

## for문 range()함수

```python
for i in range(0, 10, 1): #시작값, 끝값, 증가량 인데 끝값은-1을 뺀다 프로그래밍 언어의 입장에서는 보통 0부터 시작. 너무 당연해서 생략해도 된다
    print(i)    
# 그래서 for i in range(10)이렇게 해도 똑같이 나온다. 1증가도 당연하기에


#특정 리스트의 길이만큼 반복하고 싶다?
# 만약 Long_Lst의 길이가 35000개라면?
#i는 index를 의미하기에 통상적으로 많이 쓰인다.
#for i in range(len(long_lst)):   #멜론차트 100개중 노래제목 가수이름을 하나씩 뽑고 싶은데 같은 갯수니까 길이는 하나만 써도 된
```

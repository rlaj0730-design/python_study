## 사용자 정의 함수

내장된 함수가 아닌 내가 직접 함수를 만드는거임

def = define : 정의를 내리다

함수명이 믹서기. 매개변수가 과일. 실행문자는 어떻게 할지 설명(갈아주는거). 반환변수는 결과물. 쥬스

```python
def number_sum(num1, num2):           #숫자 두개 받아서 더하는 믹서기 만들기
    result = num1 + num2             #실행문장을 리절트에 담아주기. 쓰기 쉽게
    return result                    #예시는 쉬운 더하기로 했지만 복잡한 계산 등에 사용하면 편리하다
```

```python
number_sum(3, 10)       #믹서기에 재료넣기
```

```python
def number_minus(num1, num2):
    result = num1 - num2
    return result                    #믹서기 만들고
```

```python
num1 = int(input("첫 번째 정수 입력 >>"))
num2 = int(input("두 번째 정수 입력 >>"))
result = number_minus(num1, num2)             #믹서기 사용
result
```

```python
def number_minus(num1, num2):
    if num1 >= num2:
        result = num1 - num2
    else:
        result = num2 - num1           #큰숫자에서 작은 숫자 빼는 믹서기로 바꾸기
    return result         
```

```python
num1 = int(input("첫 번째 정수 입력 >>"))
num2 = int(input("두 번째 정수 입력 >>"))
result = number_minus(num1, num2)             
result
```

```python
def cal(num1, num2, op):
    if op == "+":
        result = num1 + num2
    

    elif op == "-":
        result = num1 - num2

    else:
        result = "잘못된 연산자 입니다"
        
    
    return result

    
```

```python
num1 = int(input("첫 번째 정수 입력 >> "))
num2 = int(input("두 번째 정수 입력 >> "))
op = input("연산자 입력(+,-) >>")
result = cal(num1, num2, op)
print(f"결과 : {result}")
```

```python
def sum_mul(sik, num1, num2, num3, num4):
    if sik == "sum":
        result = num1 + num2 + num3 + num4

    elif sik == "mul":
        result = num1 * num2 * num3 * num4

    else:
        result = "잘못된 입력 입니다"

    return result
```

```python
sik = input("sum또는 mul을 입력하시오 >>")
num1 = int(input("첫번째 숫자를 입력하시오 :"))
num2 = int(input("두번째 숫자를 입력하시오 :"))
num3 = int(input("세번째 숫자를 입력하시오 :"))
num4 = int(input("네번째 숫자를 입력하시오 :"))

result = sum_mul(sik, num1, num2, num3, num4)
print(result)                  #이거 아님 내가 푼 풀이
```

```python
def sum_mul(option, a, b, c, d):
    if option == "mul":
        return a*b*c*d
    else:
        return a+b+c+d   #선생님이 푼 풀이
    
```

```python
sum_mul('sum', 1,2,3,4)
```

```python
sum_mul('mul',1,2,3,4)
```

```python
def sum_mul(option, *args):    #args = [1,2,3,4] -> 리스트 형태로
    if option == "mul":
        result = 1
        for i in args:
            result *= i
        return result
    else:
        result = 0
        for i in args:
            result += i
        return result  #뒤에 숫자 여러개 넣고 싶을때
```

```python
sum_mul('sum', 1,2,3,4,5,6)
```

```python
sum_mul('mul',1,2,3,4,5,6)
```

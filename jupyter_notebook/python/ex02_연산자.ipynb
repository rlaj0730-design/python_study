## 산술연산자

```python
num1 = 10
num2 = 7


print(f"나누기 : {num1/num2}")
print(f"몫 : {num1//num2}")
print(f"나머지 : {num1%num2}")
```

문자열을 더할때는 +가 더이상 산술연산자가 아닌 연결 연산자 이다. 큰따옴표 안에 숫자 넣으면 문자로 판단(인풋으로 판단)

```python
# 문자의 곱셈도 가능하다!
print("=" * 10)
```

```python
num1 = 23
num2 = 3
print(f"더하기 결과 : {num1 + num2}")
print(f"빼기 결과 : {num1 - num2}")
print(f"곱하기 결과 : {num1 * num2}")
print(f"나누기 결과 : {num1 / num2}")
```

```python
num1 = int(input("num1을 입력하세요 >>"))
num2 = int(input("num2를 입력하세요 >>"))
print(f"더하기 결과 : {num1 + num2}")
print(f"빼기 결과 : {num1 - num2}")
print(f"곱하기 결과 : {num1 * num2}")
print(f"나누기 결과 : {num1 / num2}")

#input() 함수 사용 시 주의사항
# 1. input으로 들어온 값은 기본적으로 문자열이다! (숫자여도 인푹으로 되어있으면 문자로 판단.)
# 2. input 안에 들어갈 값을 모두 입력하고 다음으로 넘어가야 함.
```

input은 들어가는 값을 의미함
형변환(인풋값은 문자라 인트값.숫자로 바꿔줘야함)(int로 인풋값을 감싸주기)

인트값 마저 넣지 않고 밑에 셀에 코드 작성하면 에러가 뜬다.
그러면 연산이 끝났는지 안끝났는지 어떻게 알수있냐면 왼쪽 항목록에 숫자 형태로 뜬다([5])
([*])은 연산이 덜 끝났다는 의미. 이럴땐 실행중인 셀이 있는게 아닌지 위를 확인해볼 필요가 있다.

인풋 사용법은 꼭 기억하자!!

```python
#예제3 
time = int(input("시간 입력 >> "))
hour = time // 3600
minute = (time % 3600) // 60
second = time % 60
print(f"{hour}시간 {minute}분 {second}초") #print( "{}시간 {}분 {}초".format(hour, minute, second)) ->이거는 옛날에 쓰던 방법
```

## 복합 대입 연산자

```python
num = 27
num += 3  # num = num + 3(num값을 3을 더한 값으로 덮어 씌워주겠다는 뜻. 그래서 계속하면 최종 값에 계속 3씩 더한값이 저장됨)
print(num)
num += 3
print(num)
```

## 비교 연산자(자주 쓰임. 중요)

```python
a = 3
b = 7
print(a > b) #a가 b보다 큽니까?
print(a <= b)
print(a == b)
print(a != b) #같은게 아닌가?
```

비교 연산자를 쓰면 무조건 true or false만 나와야함
비교연산자에서는 수식의 같다 표현은 "=="이다. "="아님.
파이썬에서 느낌표는 아니다라는 뜻.

## 논리 연산자(true, false를 정해주는 연산자)

```python
#not은 논리값을 뒤집는 역할이다.
a = 3
b = 7
print(not a > b) 
print(not a <= b)
print(not a == b)
print(not a != b) 
```

```python
#and는 두 값이 모두 true일 경우에만 true이다
3 >= 5 and 10==10
```

```python
#or은 두 값중 하나라도 true이면 true이다.
3 < 5 or 10 == 20
```

## 삼항 연산자( a if 조건식 else b)(조건식 값은 참 아니면 거짓뿐)

조건식이 트루면 a가 출력되고 폴스면 b가 출력된다

```python
score = 80
"합격" if score >= 60 else "불합격"
```

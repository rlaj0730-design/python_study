컴퓨터는 숫자 0부터 세고 대괄호는 인덱스 추출해내는

글의 순서. 숫자를 인덱스라고 한다. 목차, 색인(구분한다). 숫자값이라고 생각하면 된다

띄어쓰기한 빈칸도 숫자를 세야한다

인덱스는 뒤에서 숫자를 세는 방법도 있다.

```python
name = "My name is JY"
# 대문자 M을 꺼내고 싶다면?
# 소문자 m을 꺼내고 싶다면?
# 대문자 y를 꺼내고 싶다면?
print(name[0])
print(name[3])
print(name[-1])
```

뒤에서 인덱싱(추출)을 할땐 0이 앞에 쓰였기 때문에 -1부터 숫자를 센다

슬라이싱은 하나의 문자를 잘라서 보겠습니다 라는 뜻

### 슬라이싱

```python
name = "My name is JY"
# MY를 추출하고 싶을때
name[0:2]
```

슬라이싱을 할땐 내가 출력하고자 하는 인덱스 값 더하기 1을 해야한다. 그리고 :은 ~부터 라는 뜻.

```python
name = "My name is JY"
# name 꺼내기
print(name[3:7]) # 3 <= x <7
# is JY를 꺼내고 싶다!
print(name[8:])
# My name을 꺼내고 싶다!
print(name[:7])
# 처음 부터 끝까지 다 꺼내고 싶다
print(name[:])
```

```python
s = "2026년 8월 6일의 날씨는 맑음입니다."
print("날짜 :", s[:11])
print("날씨 :", s[-6:-4])
```

## numpy 라이브러리 사용법

```python
# numpy 라이브러리 로딩(불러오기)
import numpy as np
#as --> alias의 줄임말. numpy라이브러리를 앞으로 np라고 부르겠다!
```

#### 리스트
- 여러가지 자료형을 가질 수 있다.
- 메모리 용량이 크고 속도가 매우 느리다.

#### numpy
- 동일한 자료형만 가지는 배열
- 가볍고 계산속도가 우월하다
- n차원 형태로 구성이 가능하다  이미지, 동영상에 필수.

```python
# array 생성해보기
lst = [1, 2, 3, 4, 5]

arr = np.array(lst)
arr
```

```python
# 2차원 array 생성
arr2 = np.array([[1,2,3],
                 [4,5,6]])   #문 두번 열면 2차원을 생성한다는 뜻
arr2
```

#### array 클래스의 속성들을 사용해서 자료구조 파악하기

```python
# 배열의 크기(모양)을 확인하는 방법 : shape*************매우매우중요함
print(arr.shape)     # (5,) --> 1차원 배열 --> 데이터의 갯수 : 5개
print(arr2.shape)     #(2,3) --> 2차원 배열 --> (행, 열)
# 쉐잎은 소괄호 없이 사용한다!! 엄청 중요
```

```python
#배열이 가진 요소의 갯수 확인 : size
print(arr.size) # --->5개
print(arr2.size) # ---> 2*3   6개
```

```python
# 배열의 차원 확인 : ndim
print(arr.ndim)
print(arr2.ndim)
```

```python
arr3 = np.array([[[1,2],
                 [3,4]],

                [[5,6],
                 [7,8]]])
```

```python
print("배열의 크기 :", arr3.shape)
print("배열의 차원 :", arr3.ndim)
print("배열의 개수 :", arr3.size)
```

#### 특정 숫자로 배열 채우기

```python
# np.zeros(배열의 크기)
np.zeros((2, 2, 2))
```

```python
# 특정 숫자로 배열 채우기
np.full((2,2,2),7)
```

```python
# 하나씩 증가하는 숫자로 배열 채우기
np.arange(1,51)   #array range
```

```python
# 랜덤한 숫자로 배열 채우기
# numpy에서는 random 라이브러리를 불러오지 않고도 랜덤한 값을 뽑아올 수 있다!!!
# 대신, 끝 값이 포함 안된다
np.random.randint(1,3)
```

```python
# 랜덤 값으로 array 만들기
np.random.randint(1,11,size = (2,3))
```

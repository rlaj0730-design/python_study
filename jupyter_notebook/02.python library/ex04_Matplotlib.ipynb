### Matplotlib 라이브러리 사용법

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 데이터 분석에 필요한 라이브러리 3대장
import seaborn as sns # 최근에 추가됨. 데이터를 던져주면 알아서 그려주는 도구임. 현대적인 그래프가 많음
```

```python
#기본적인 그래프 그리기 - 선 그래프
y = [2, 12, 6]
plt.plot(y)    #값 하나만 넣으면 무조건 y축에 우선으로 배치됨

# plot(좌표 y요소) --> x좌표는 0부터 1씩 증가하며 자동으로 할단된다

plt.show() #print가능
```

```python
# x축과 y축을 설정해서 시각화
x = [1, 2, 3]
y = [2, 4, 6]

plt.plot(x,y,ls = "-.")  #ls = line style
plt.show()
```

```python
x = [1, 2, 3]
y = [2, 4, 6]

plt.plot(x,y,ls = "--", marker = "o")  #x축 y축 값 만나는 지점 마킹해주기. $★$도 가능
plt.show()
```

```python
x = [1, 2, 3]
y = [2, 4, 6]

plt.plot(x,y,ls = "--", marker = "D", lw = 5)  #lw = 라인두께 설정
plt.show()
```

```python
x = [1, 2, 3]
y = [2, 4, 6]

plt.plot(x,y,ls = "--", marker = "D", lw = 5, color = "green") #css색상정보 코드를 넣거나 컬러 영어 이름넣어도 됨
plt.show()
```

```python
# 속성값
# 1. 라인의 스타일 설정 : ls(Line Style) -> 문자열
# 2. 라인의 두께 지정 : lw(Line Width) -> 그냥 숫자로 적어주기!! 문자X
# 3. 선의 색 바꾸기 : c(color) -> 선색과 마커의 색까지 모두 변경
# 4. 값의 위치 표시 : marker -> 문자열
# 5. 마커 내부 색깔 : mfc(marker face color) -> 문자열
# 6. 마커 선 색깔 : mec(marker edge color) -> 문자열
# 7. 마커 크기 : ms -> 숫자
# 8. 마커 선 굵기 : mew(marker egde width) -> 숫자
```

```python
x = np.arange(8)
y = [5,2,3,6,8,7,5,7]

plt.plot(x,y,ls = "--", marker = "o", color = "green")   #컬러 줄여서 c만 적어도 되고 색상도 앞자만 적어도 됨. rgb
plt.show
```

```python
x = np.arange(8)
y = [5,2,3,6,8,7,5,7]

plt.plot(x,y, ls = ":", marker = "$◆$", mfc = "blue", mec = "red", color = "red", ms = 5) #마커는 대문자 D로 하면 된다
plt.show
```

```python
from matplotlib.offsetbox import AnnotationBbox, OffsetImage
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(8)
y = [5, 2, 3, 6, 8, 7, 5, 7]


colors = plt.cm.rainbow(np.linspace(0, 1, len(x) - 1))


fig, ax = plt.subplots(facecolor="#FFF8DC")
ax.set_facecolor("#FFF8DC")


for i in range(len(x) - 1):
    ax.plot(
        x[i : i + 2],
        y[i : i + 2],
        color=colors[i],
        ls=":",
        lw=2,
        marker="$★$",
        ms=25,
        mfc="yellow",
        mec="gold",
    )


try:
    cat_img = plt.imread("cat.jpg")
    for xi, yi in zip(x, y):
        imagebox = OffsetImage(cat_img, zoom=0.1)
        ab = AnnotationBbox(imagebox, (xi, yi), frameon=False)
        ax.add_artist(ab)
except FileNotFoundError:
    print("'cat.jpg' 파일을 찾을 수 없습니다.")

plt.tight_layout()
plt.show()
```

```python
x = np.arange(8)
y = [5,2,3,6,8,7,5,7]

plt.plot(x,y)

#범위를 인위적으로 조작하여 그래프의 크기를 조절할 수 있다
plt.xlim(-2, 10)  # x,y값을 내가 설정 하겠습니다 
plt.ylim(-5, 12)
plt.show()
```

```python
x = np.arange(8)
y = [5,2,3,6,8,7,5,7]

plt.plot(x,y)

# 그리드 그리기 -> 격자눈금, 모눈종이
plt.grid()

# 눈금 출력 방식 -> ticks   #y값에 해당하는 숫자가 있으면 y축에 숫자를 보여주고 없으면 안보여줌
plt.yticks(y)

plt.show()
```

```python
# 하나의 차트에 여러 데이터를 입력하고 한 번에 그리기
x = [1,2,3,4]
y = [2,4,6,8]
z = [3,6,9,12]

plt.plot(x, label = "x")
plt.plot(y, label = "y")
plt.plot(z, label = "z")

plt.legend()  # 범례표시하는 도구

# 그래프 제목
plt.title("차트", loc = "right") # -> left, center, right 다양하게 사용 가능

plt.xlabel("x축")
plt.ylabel("y축")

plt.show()
```

```python
# 실습 : 1~11 사이의 랜덤한 숫자를 y1와 y2에 저장하고(x값은 고정)
# y1와 y2의 선 그래프를 하나의 그래프에 같이 표현되도록 그려보자
# np.random.randint(1, 11, size = 7), x = range(2, 9)
x = range(2,9)
y1 = np.random.randint(1, 11, size = 7) # 내가 가진 데이터의 범위 값
y2 = np.random.randint(1, 11, size = 7) 

plt.plot(x,y1, color = "purple")    #플롯은 값하나만 적으면 y축이 됨. x는 자동 생성
plt.plot(x,y2, color ="g")
plt.legend()
plt.grid()
plt.title("랜덤 값 시각화", loc = "right")

plt.show()
```

### 바 그래프 그리기

```python
x = [0,1,2,3,4,5]
y = [80,95,70,60,50,90]

plt.bar(x,y)
plt.xlabel("소속반")
plt.ylabel("점수")

plt.show()
```

```python
x = [0,1,2,3,4,5]
y = [80,95,70,60,50,90]

plt.barh(x,y, height = 0.5)   #bar뒤에 "h"를 붙여주면 가로로 생성이 된다 horizontal(수평의)  #bar굵기도 조절가능. 세로바는 width
plt.xlabel("소속반")
plt.ylabel("점수")            #데이터간의 크기를 비교해서 보고싶을때 바그래프를 쓴다. 선그래프는 변화율.추이를 볼때 쓴다
plt.xlim(40,100)
                              
plt.show()
```

#### 데이터 설명
- 0행은 키 정보, 1행은 몸무게 정보가 담겨있다
- 몸무게는 pound 단위로 되어있고, 키는 cm 단위로 되어있다
- 그런데 우리가구하려는 BMI 지수는 kg과 m가 필요하다 --> 단위 변환을 해줘야 한다

```python
# 데이터 생김새 --> 크기
data.shape
```

```python
# 1. 키 데이터 가져오기(인덱싱) --> h라는 변수에 담아주기
# 2. 몸무게 데이터 가져오기(인덱싱) --> w라는 변수에 담아주기
# 3. 키 데이터를 cm -> m로 변환
# cm * 0.01 --> m
# 4. 몸무게 데이터는 pound -> kg으로 단위변화
#    pound / 2.204--> kg
# 5. bmi 지수 구하는 공식을 적용!
#     몸무게(kg) / 키(m) * 키(m)
```

```python
h = data[0, 0:10]
w = data[1, 0:10]

bmi = (w / 2.204) / ((h * 0.01)*(h * 0.01))

bmi    #내가 푼 풀이
```

```python
h = data[0]
w = data[1]

h = h * 0.01    #h와 w에 다시 값 넣어주기
w = w / 2.204

bmi = w / (h * h)
bmi   #선생님이 푼 풀이
```

### numpy Boolean 인덱싱

데이터 분석의 꽃이라 불릴정도로 중요중요함. 트루 폴스로 인덱싱하는거

```python
score = np.array([80, 75, 62, 37, 90])
score


# 예를 들어, score가 100만개 있다고 가정
```

```python
# 합격 기준이 70점 이상!  #true와 false의 집합은 재료다! 매우매우 중요
score >= 70    #요소별 연산이 가능하기 때문에  70점 이상인지 점수를 다 대조함 
```

트루랑 폴스가 모여있는 집합을 보면 재료라고 부르면 됩니다!

```python
score[score >= 70]    #score >= 가 재료라고 보면 됨. 원래는 조건문 반복문 사용해서 일일히 대조했어야함(if문 for문)
                      # 근데 불리안 인덱싱 한줄로 바로 정리할수있고 연산속도도 빠름
```

```python
score2 = np.array([[20, 90, 40], [50, 30, 90]])
score2
```

```python
score2[score2 >= 70]
```

### numpy 최종 실습

```python
# 데이터 불러오기 --> 영화 리뷰 평점 데이터 분석
movie = np.loadtxt("./data/ratings.txt", delimiter = ",", dtype = "int64")
movie
```

```python
# 데이터를 불러오고 나면 항상 데이터의 모양을 확인해야한다!
movie.shape    #데이터 100만개


#데이터의 열(컬럼) 이름
#사용자 아이디 // 영화 아이디 // 영화 평점 // 리뷰 입력 시간
```

#### 전체 영화 평점의 평균

```python
all_movie = movie[:, 2]
all_movie   # 전체 영화 평점의 --->평균을 구하면 --> 전체 영화 평점이 나옴!
np.mean(all_movie)  #대부분 사람들은 영화를 보고 좋든 싫든 70점 정도는 주는구나 하고 데이터 분석을 할수 있음.
```

#### 1번 사용자가 매긴 평점의 평균 구해보기

```python
# 1. 0번 인덱스 열에 있는 데이터가 1번 사용자인지 아닌지 확인해야합니다 ----> boolean인덱싱
# 2. 1번 사용자의 평점만 모아서 평점 평균 구해주기
people = movie[:, 0]        #내가 푼 풀이.

np.mean(all_movie[people == 1]) # 1에 큰따옴표 지워야하는 이유는 넘파이 파일은 다 산술 숫자로 이루어졌기 때문
```

```python
# 1. 0번 인덱스 열에 있는 데이터가 1번 사용자인지 아닌지! --> boolean인덱싱
movie[:,0] == 1  # 재료
movie_2 = movie[movie[:,0] == 1]   # 1번 사용자가 매긴 평점 전체
# 2. 1번 사용자의 평점만 모아서 평균 구해주기
np.mean(movie_2[:, 2])
```

### 각 사용자가 매긴 평점의 평균을 모두 구해보자

```python
# 일단, 사용자들을 중복없이 정리한 리스트가 필요하다
# [1,2,3,4,......, 6039, 6040]
movie[:,0]

user_id = np.unique(movie[:, 0]) #중복 없애는 함수
user_id
```

```python
# 반복문을 사용해서 각 사용자의 평점의 평균을 구해보자
user_mean_lst = []
movie_2 = movie[movie[:, 0] == 1]  
np.mean(movie_2[:, 2])          #1번 사용자의 평점의 평균
    
    
```

```python
for i in user_id: #6040번 반복
    movie_2 = movie[movie[:, 0] == i] 
    user_mean_lst.append([i, np.mean(movie_2[:, 2])])
user_mean_lst
    
```

#### 각 사용자들의 평점의 평균이 4점 이상인 평점만 구해보자

```python
type(user_mean_lst)  # --> 리스트는 요소별 연산이 불가능! --> array로 바꿔줘야함
user_mean_arr = np.array(user_mean_lst)
user_mean_arr

# e+00 --> 컴퓨터의 지수 표현 방식이다.
# 아주 작은 값이나 큰 값을 표현할 때 10의 지수로 근사치를 표현하는 방식이다.
```

```python
# 실수를 정수로 바꿔보자
user_mean_arr = user_mean_arr.astype(np.int64)    #astype = 타입을 바꾸겠다
user_mean_arr[:, 1] >= 4
```

```python
user_mean_arr[user_mean_arr[:, 1] >= 4]
```
